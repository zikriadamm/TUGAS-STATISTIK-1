```python
# import
from sklearn.linear_model import LinearRegression
import pandas as pd
import matplotlib.pyplot as plt
```

```python
model = LinearRegression()

df = pd.read_csv("../dataset/data_nilai.csv")

x = df[["jam_belajar"]]
y = df["nilai_ujian"]


# train
model.fit(x, y)
```

```python
new_data = pd.DataFrame([[15]], columns=["jam_belajar"])

# prediction
prediction = model.predict(new_data)

print(f"prediksi nilai ujian dengan 15 jam belajar {prediction}")
```

```python
# visualisasi
plt.scatter(x, y)

plt.plot(
    x,
    model.predict(x),
)

plt.xlabel("jam belajar")
plt.ylabel("nilai ujian")
plt.title("Linear Regression - Jam Belajar vs Nilai Ujian")

plt.show()
```
